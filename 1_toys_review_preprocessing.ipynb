Review data inspection

```python
reviews_raw = reviews.copy()

print("Dataset shape:", reviews_raw.shape)
print("\nColumn names:")
print(reviews_raw.columns.tolist())

print("\nData types:")
print(reviews_raw.dtypes)

print("\nMissing values:")
print(
    reviews_raw.isna()
    .sum()
    .sort_values(ascending=False)
)
```

Keep relevant cols

```python
keep_columns = [
    "rating",
    "title",
    "text",
    "parent_asin",
    "timestamp",
    "helpful_vote",
    "verified_purchase", 
    "user_id"
]

# Only select columns that actually exist
keep_columns = [
    col for col in keep_columns
    if col in reviews_raw.columns
]

reviews_selected = reviews_raw[keep_columns].copy()

print("Selected dataset shape:", reviews_selected.shape)
```

```python
# Check data types and missing values

# Check ratings are within the expected range
print("Rating values:", sorted(reviews_selected["rating"].unique()))
print("Invalid ratings:", (~reviews_selected["rating"].between(1, 5)).sum())

# Check for empty or whitespace-only review text
empty_title = reviews_selected["title"].str.strip().eq("")
empty_text = reviews_selected["text"].str.strip().eq("")

print("Empty titles:", empty_title.sum())
print("Empty review bodies:", empty_text.sum())

# Keep reviews with a valid rating and at least a title or body
reviews_selected = reviews_selected[
    reviews_selected["rating"].between(1, 5)
    & ~(empty_title & empty_text)
].copy()

print("Cleaned dataset shape:", reviews_selected.shape)
```

```python
reviews_selected["review_text"] = (
    reviews_selected["title"].fillna("")
    + ". "
    + reviews_selected["text"].fillna("")
).str.strip()


# Remove rows with no usable review text
reviews_selected = reviews_selected[
    reviews_selected["review_text"].ne("")
].copy()

print(
    reviews_selected[["title", "text", "review_text"]]
    .head(10)
    .to_string(index=False)
)
```

Create sentiment labels from ratings

```python
def rating_to_sentiment(rating):
    if rating >= 4:
        return "positive"
    elif rating == 3:
        return "neutral"
    else:
        return "negative"

reviews_selected["sentiment"] = reviews_selected["rating"].apply(rating_to_sentiment)

print(reviews_selected[["rating", "sentiment"]].head(10))

print("\nSentiment counts:")
print(reviews_selected["sentiment"].value_counts())

print("\nSentiment proportions:")
print(
    reviews_selected["sentiment"]
    .value_counts(normalize=True)
    .mul(100)
    .round(2)
)
```

```python
# Check duplicate rows

print(
    "Exact duplicate rows:",
    reviews_selected.duplicated().sum()
)

print(
    "Repeated review texts:",
    reviews_selected["review_text"].duplicated().sum()
)

duplicates = reviews_selected[reviews_selected.duplicated(keep=False)]

print("Rows involved in exact duplicates:", len(duplicates))
display(duplicates.head(20))
```

```python
# Count occurrences of each review text
text_counts = reviews_selected["review_text"].value_counts()

# Keep only texts appearing more than once
repeated_texts = text_counts[text_counts > 1]

print("Number of distinct repeated texts:", len(repeated_texts))
print("Total rows containing repeated text:",
      repeated_texts.sum())

# Show the 20 most frequently repeated texts
print(repeated_texts.head(20))

```

```python

# Count occurrences of each review text + user ID combination duplicates
text_user_counts = (
    reviews_selected.groupby(["review_text", "user_id"])
    .size()
    .sort_values(ascending=False)
)

# Keep only combinations appearing more than once
repeated_text_user = text_user_counts[text_user_counts > 1]

print("Distinct repeated text-user combinations:",
      len(repeated_text_user))

print("Total rows containing repeated combinations:",
      repeated_text_user.sum())


# Get all repeated text-user combinations
repeated_pairs = repeated_text_user.index

# Display the matching reviews and metadata
repeated_reviews = (
    reviews_selected.set_index(["review_text", "user_id"])
    .loc[repeated_pairs]
    .reset_index()
)

display(
    repeated_reviews[
        [
            "review_text",
            "user_id",
            "parent_asin",
            "rating",
            "timestamp"
        ]
    ].sort_values(["user_id", "timestamp"])
)

```

```python
# remove exact duplicate rows
reviews_selected = reviews_selected.drop_duplicates(keep="first").copy()

print("Dataset shape after removing exact duplicates:", reviews_selected.shape)
```

```python
# remove duplicate rows based on review_text and same user_id, posted within 24 hrs of each other, keeping the first occurrence

# Ensure timestamps are in datetime format
reviews_selected["timestamp"] = pd.to_datetime(
    reviews_selected["timestamp"]
)

# Sort reviews chronologically within each user and text group
reviews_selected = reviews_selected.sort_values(
    ["user_id", "review_text", "timestamp"]
).copy()

# Identify reviews to keep
keep_indices = []

for _, group in reviews_selected.groupby(
    ["user_id", "review_text"],
    sort=False,
    dropna=False
):
    last_kept_time = None

    for idx, row in group.iterrows():
        current_time = row["timestamp"]

        if pd.isna(current_time):
            # Keep reviews with missing timestamps
            keep_indices.append(idx)
            continue

        if (
            last_kept_time is None
            or current_time - last_kept_time
            > pd.Timedelta(hours=24)
        ):
            keep_indices.append(idx)
            last_kept_time = current_time

# Keep the selected rows
before = len(reviews_selected)

reviews_selected = reviews_selected.loc[
    keep_indices
].sort_index().copy()

print("Reviews before:", before)
print("Reviews after:", len(reviews_selected))
print("Reviews removed:", before - len(reviews_selected))
```

```python
# Count each review text + user ID combination
duplicate_counts = (
    reviews_selected
    .groupby(["review_text", "user_id"], dropna=False)
    .size()
    .reset_index(name="duplicate_count")
)

# Keep only repeated combinations
duplicate_counts = duplicate_counts[
    duplicate_counts["duplicate_count"] > 1
].sort_values("duplicate_count", ascending=False)

print("Repeated text-user combinations:", len(duplicate_counts))
print("Rows belonging to repeated combinations:",
      duplicate_counts["duplicate_count"].sum())

display(duplicate_counts.head(20))
```

```python
# Get all rows belonging to repeated text-user combinations
duplicate_reviews = reviews_selected.merge(
    duplicate_counts,
    on=["review_text", "user_id"],
    how="inner"
)

# Select available columns for inspection
inspect_cols = [
    col for col in [
        "review_text",
        "user_id",
        "duplicate_count",
        "parent_asin",
        "rating",
        "timestamp"
    ]
    if col in duplicate_reviews.columns
]

# Sort by user, text, and time
duplicate_reviews = duplicate_reviews.sort_values(
    ["user_id", "review_text", "timestamp"]
)

display(duplicate_reviews[inspect_cols].head(20))
```
